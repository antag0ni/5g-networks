```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc

EbN0_dB = np.arange(0, 35.5, 0.5)
g = 10 ** (EbN0_dB / 10)                      # Eb/N0 lineare

ber_awgn = 0.5 * erfc(np.sqrt(g))             # Q(sqrt(2 Eb/N0))
ber_rayl = 0.5 * (1 - np.sqrt(g / (1 + g)))   # fading Rayleigh piatto, rivelazione coerente

plt.figure(figsize=(8, 5.5))
plt.semilogy(EbN0_dB, ber_awgn, 'b-', lw=2, label='AWGN')
plt.semilogy(EbN0_dB, ber_rayl, color='orange', ls='-', lw=2, label='Fading')
plt.grid(True, which='both', ls=':', alpha=0.6)
plt.ylim(1e-6, 1)
plt.xlabel('$E_b/N_0$ [dB]')
plt.ylabel('BER')
plt.title('2-PAM: Solo AWGN vs Con Fading')
plt.legend()
plt.tight_layout()
plt.savefig('ber_2pam.png', dpi=150)
```
